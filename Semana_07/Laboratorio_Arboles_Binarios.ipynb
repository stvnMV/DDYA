## 1) Descripción del problema **[1 pts]**

Una empresa necesita organizar los códigos numéricos de los productos almacenados en una bodega. Los códigos son ingresados por el usuario en un orden determinado y deben almacenarse inicialmente en una lista.


El sistema debe utilizar los mismos códigos para construir dos estructuras diferentes: un Árbol Binario de Búsqueda (BST) y un Árbol AVL.


El BST debe conservar exactamente el orden de inserción y no realizar ningún tipo de balanceo. En cambio, el AVL debe revisar el factor de balance después de cada inserción y realizar las rotaciones necesarias cuando algún nodo quede desbalanceado.


Finalmente, el programa debe permitir realizar los recorridos preorden, inorden y postorden sobre ambos árboles y mostrar gráficamente la estructura obtenida para poder comparar su organización.
