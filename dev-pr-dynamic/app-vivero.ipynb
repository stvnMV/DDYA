## 6) Código funcional **[15 pts]**



```python
memo={}
def compr(a):
    try:
        val = int(a)

        if val < 0:
            raise ValueError
        return val,1

    except ValueError:
        print("Dato invalido, por favor asegurese de ingresar solo: \n- Numeros enteros \n- Numeros positivos \n- No dejar el campo vacio\n- No ingresar letras ni simbolos\n- Introducir un solo numero\n- No usar espacios, comas o cualquier otro caracter")
        return 0,0

def compr_test(a):
    try:
        val = int(a)
        if val < 0:
            raise ValueError
        return 1

    except ValueError:
        print("Dato invalido, por favor asegurese de ingresar solo: \n- Numeros enteros \n- Numeros positivos \n- No dejar el campo vacio\n- No ingresar letras ni simbolos\n- Introducir un solo numero\n- No usar espacios, comas o cualquier otro caracter")
        return 0

def plant(n):
    if n in memo:
        return memo[n]
    elif n==0:
        memo[0]=0
        return n
    elif n==1:
        memo[1]=1
        return n
    else:
        num_plant=(plant(n-1)+plant(n-2))
        memo[n]=num_plant
        return num_plant

def test(n,m):
    b=compr_test(n)
    if b==0:
        return
    b=compr_test(m)
    if b==0:
        return
    if n in memo:
      print("El numero de plantas en el mes",n,"es de:",memo[n])
      if m==memo[n]:
        return print("la salida corresponde a al resultado ")
      else:
        return print("la salida no corresponde a al resultado ")
    elif n==0:
        memo[0]=0
        return n
    elif n==1:
        memo[1]=1
        return n
    else:
        num_plant=(plant(n-1)+plant(n-2))
        memo[n]=num_plant
        print("El numero de plantas en el mes",n,"es de",num_plant)
        if m==num_plant:
            return print("la salida corresponde a al resultado ")
        else:
            return print("la salida no corresponde a al resultado ")

def main():
    ctrl=2
    while ctrl==2:
        b=0
        while b==0:
            n=input("Ingrese el mes en el que quiere saber las plantas producidas hasta entonces:")
            n,b=compr(n)
        print("El numero de plantas en el mes",n,"es:",plant(n))
        b=0
        while b==0:
            ctrl=input("1) presione 1 si quiere terminar el programa\n2) presione 2 si desea seguir buscando\n__")
            ctrl,b=compr(ctrl)
    print("GRACIAS")
main()
```
